```python
df["Hour"] = df["Datetime"].dt.hour

df["Day"] = df["Datetime"].dt.day

df["Month"] = df["Datetime"].dt.month

df["DayOfWeek"] = df["Datetime"].dt.dayofweek

df["Week"] = df["Datetime"].dt.isocalendar().week

df["Quarter"] = df["Datetime"].dt.quarter

df["Weekend"] = df["DayOfWeek"].isin([5, 6]).astype(int)
```

```python
df.head()
```

```python
df["Lag_1"] = df["Global_active_power"].shift(1)
```

```python
df["Lag_24"] = df["Global_active_power"].shift(24)
```

```python
df["Lag_168"] = df["Global_active_power"].shift(168)
```

```python
df["RollingMean_24"] = (
    df["Global_active_power"]
    .rolling(window=24)
    .mean()
)

df["RollingStd_24"] = (
    df["Global_active_power"]
    .rolling(window=24)
    .std()
)
```

```python
df["Difference"] = df["Global_active_power"].diff()

df["PctChange"] = df["Global_active_power"].pct_change()
```

```python
threshold = df["Global_active_power"].quantile(0.75)

df["Peak"] = (
    df["Global_active_power"] > threshold
).astype(int)
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

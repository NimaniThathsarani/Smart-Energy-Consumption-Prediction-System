# 03 - Data Pipeline and Validation

This notebook validates the Member 1 intermediate dataset and creates the final processed output.

```python
# Import the reusable Member 2 pipeline.
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / 'src').exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

from src.preprocessing import load_intermediate_data
from src.validation import validate_dataframe, quality_check_passed
```

```python
# Load and standardize the Member 1 intermediate CSV.
input_path = project_root / 'data/interim/energy_cleaned_intermediate.csv'
df = load_intermediate_data(input_path)
print('Shape:', df.shape)
print('Timestamp dtype:', df['Timestamp'].dtype)
```

```python
# Run the complete schema, timestamp, and measurement validation.
summary = validate_dataframe(df)
print('Quality checks passed:', quality_check_passed(summary))
print('Timestamp checks:', summary['timestamp'])
print('Measurement missing values:', {column: details['missing_values'] for column, details in summary['measurements'].items()})
```

```python
# Create the final processed CSV after validation succeeds.
output_path = project_root / 'data/processed/energy_clean.csv'
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False, date_format='%d/%m/%Y %H:%M:%S')
print('Saved:', output_path)
```
